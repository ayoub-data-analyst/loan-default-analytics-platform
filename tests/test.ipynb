```python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
pd.set_option("display.width", None)
pd.set_option("display.max_colwidth", None)

df = pd.read_parquet("../dataset/rejected_2007_to_2018Q4.parquet")
```

```python

# Shape
print(df.shape)

# Columns
print(df.columns.tolist())

# Data types
print(df.dtypes)

# Info
df.info()

# Missing values
missing = pd.DataFrame({
    "Null": df.isnull().sum(),
    "Null %": (df.isnull().mean()*100).round(2)
})

print(missing.sort_values("Null %", ascending=False))

# Unique values
unique = df.nunique().sort_values()

print(unique)

# Numerical statistics
print(df.describe().T)

# Categorical statistics
print(df.describe(include="object").T)

# Duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Save reports
missing.to_csv("rejected_missing.csv")
unique.to_csv("rejected_unique.csv")
```
